```python
import pandas as pd

df = pd.read_csv("/content/daily-min-temperatures.csv")

print("Dataset loaded successfully!")
print(df.head())
print("Total records:", len(df))
print("Missing values:")
print(df.isnull().sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# Convert Date column into datetime format
df["Date"] = pd.to_datetime(df["Date"])

# Sort records by date
df = df.sort_values("Date")

# Extract temperature values
temperature = df["Temp"].values.reshape(-1, 1)

# Normalize values between 0 and 1
scaler = MinMaxScaler()
scaled_temp = scaler.fit_transform(temperature)

print("Preprocessing completed!")
print("Total records:", len(scaled_temp))
print("First 5 scaled values:")
print(scaled_temp[:5])
```

```python
# Create sequences
X = []
y = []

window_size = 30

for i in range(window_size, len(scaled_temp)):
    X.append(scaled_temp[i-window_size:i])
    y.append(scaled_temp[i])

X = np.array(X)
y = np.array(y)

# Split chronologically: 80% training, 20% testing
split = int(len(X) * 0.8)

X_train = X[:split]
y_train = y[:split]

X_test = X[split:]
y_test = y[split:]

print("Sequence creation completed!")
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout

model = Sequential([
    Input(shape=(30, 1)),
    LSTM(32),
    Dropout(0.1),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)

print("LSTM training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Predict temperatures for the test dataset
y_pred = model.predict(X_test)

# Convert scaled values back to actual temperatures
y_test_actual = scaler.inverse_transform(y_test)
y_pred_actual = scaler.inverse_transform(y_pred)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test_actual, y_pred_actual)

rmse = np.sqrt(
    mean_squared_error(y_test_actual, y_pred_actual)
)

print("Mean Absolute Error (MAE):", round(mae, 2), "°C")
print("Root Mean Squared Error (RMSE):", round(rmse, 2), "°C")
```

```python

# Get the dates for the test dataset
test_dates = df["Date"].iloc[window_size + split:]

# Plot actual and predicted temperatures
plt.figure(figsize=(14, 5))

plt.plot(
    test_dates,
    y_test_actual.flatten(),
    label="Actual Temperature"
)

plt.plot(
    test_dates,
    y_pred_actual.flatten(),
    label="Predicted Temperature"
)

plt.title("Actual vs Predicted Daily Temperatures")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
